```python
import random

numbers = [random.randint(1, 100) for i in range(10)]

print("Random numbers:")
print(numbers)

print("Even numbers:")
for number in numbers:
    if number % 2 == 0:
        print(number)
```

```python
text = "Artificial Intelligence"

print("Original string:")
print(text)

print("String in reverse order:")
print(text[::-1])

print("Words in reverse order:")
words = text.split()
print(" ".join(words[::-1]))
```
