```python
# ==========================================
# EXPERIMENT 1: CREATE CSV FILE
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re

data = {
    "Name": ["Rahul", "Priya", "Aman", "Neha", "Rohit"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 90, 78, 88, 95],
    "City": ["Delhi", "Noida", "Delhi", "Gurgaon", "Noida"]
}

df = pd.DataFrame(data)

df.to_csv("experiment_data.csv", index=False)

print("1. CSV file created successfully!")
print(df)


# ==========================================
# EXPERIMENT 2: 2-D RANDOM DATA USING NUMPY
# ==========================================

random_data = np.random.randint(1, 101, size=(10, 5))

print("\n2. 2-D Random Dataset:")
print(random_data)

print("\nShape:", random_data.shape)

print("\nRow-wise Mean:")
print(np.mean(random_data, axis=1))

print("\nColumn-wise Mean:")
print(np.mean(random_data, axis=0))

print("\nMedian:", np.median(random_data))
print("Standard Deviation:", np.std(random_data))
print("Minimum:", np.min(random_data))
print("Maximum:", np.max(random_data))


# ==========================================
# EXPERIMENT 3: PANDAS DATA CLEANING USING REGEX
# ==========================================

dirty_data = {
    "Name": ["Rahul123", "Priya@", "Aman456", "Neha", "Rohit#"],
    "Phone": ["98765-43210", "9876543211", "98-76543212", "98765 43213", "9876543214"],
    "Email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "aman@gmail.com",
        "invalid-email",
        "rohit@gmail.com"
    ]
}

clean_df = pd.DataFrame(dirty_data)

print("\n3. Original Data:")
print(clean_df)

# Remove unwanted characters from Name
clean_df["Name"] = clean_df["Name"].str.replace(
    r"[^A-Za-z ]", "", regex=True
)

# Remove non-numeric characters from Phone
clean_df["Phone"] = clean_df["Phone"].str.replace(
    r"\D", "", regex=True
)

# Check valid email
email_pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

clean_df["Valid_Email"] = clean_df["Email"].apply(
    lambda x: bool(re.match(email_pattern, x))
)

print("\nCleaned Data:")
print(clean_df)


# ==========================================
# EXPERIMENT 4: SINE WAVE USING MATPLOTLIB
# ==========================================

x = np.linspace(0, 2 * np.pi, 100)

y = np.sin(x)

plt.plot(x, y)

plt.title("Sine Wave")
plt.xlabel("X")
plt.ylabel("sin(X)")

plt.grid(True)

plt.show()
```
