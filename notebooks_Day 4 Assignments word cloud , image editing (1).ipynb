```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Text for the word cloud
text = """
Artificial Intelligence Machine Learning Deep Learning
Neural Networks Data Science Python Computer Vision
Natural Language Processing Robotics Automation
Generative AI ChatGPT Computer Vision Algorithms
Classification Regression Prediction Data Analysis
Artificial Intelligence Machine Learning Python
Deep Learning Neural Networks AI AI AI
Machine Learning Data Science Python
"""

# Create word cloud
wordcloud = WordCloud(
    width=1000,
    height=600,
    background_color="white",
    colormap="viridis"
).generate(text)

# Display word cloud
plt.figure(figsize=(12, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Artificial Intelligence Word Cloud", fontsize=20)
plt.show()

# Save as JPG
wordcloud.to_file("AI_WordCloud.jpg")

print("Word cloud saved successfully as AI_WordCloud.jpg")
```

```python

# IMAGE EDITING USING OPENCV-PYTHON




import cv2
import matplotlib.pyplot as plt
import os



image_name =  "opencv_test_image.jpg"  
# Read image
img = cv2.imread(image_name)

# Check image
if img is None:
    print("ERROR: Image not found!")
    print("Please put the image in the same folder as your notebook.")
else:

    print("Image loaded successfully!")
    print("Original image size:", img.shape)


    resized = cv2.resize(img, (500, 350))

    height, width = img.shape[:2]

    cropped = img[
        height//4:3*height//4,
        width//4:3*width//4
    ]

   

    center = (width // 2, height // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        45,
        1.0
    )

    rotated = cv2.warpAffine(
        img,
        matrix,
        (width, height)
    )

   

    flipped = cv2.flip(img, 1)


    grayscale = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2GRAY
    )

   
    blurred = cv2.GaussianBlur(
        img,
        (15, 15),
        0
    )


    bright = cv2.convertScaleAbs(
        img,
        alpha=1.3,
        beta=40
    )

    

    edges = cv2.Canny(
        grayscale,
        100,
        200
    )

  

    text_image = img.copy()

    cv2.putText(
        text_image,
        "OpenCV Image Editing",
        (30, 60),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

   

    bordered = cv2.copyMakeBorder(
        img,
        20,
        20,
        20,
        20,
        cv2.BORDER_CONSTANT,
        value=(255, 0, 0)
    )

    

    plt.figure(figsize=(16, 12))

    plt.subplot(3, 4, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original")
    plt.axis("off")

    plt.subplot(3, 4, 2)
    plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
    plt.title("Resized")
    plt.axis("off")

    plt.subplot(3, 4, 3)
    plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
    plt.title("Cropped")
    plt.axis("off")

    plt.subplot(3, 4, 4)
    plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
    plt.title("Rotated")
    plt.axis("off")

    plt.subplot(3, 4, 5)
    plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
    plt.title("Flipped")
    plt.axis("off")

    plt.subplot(3, 4, 6)
    plt.imshow(grayscale, cmap="gray")
    plt.title("Grayscale")
    plt.axis("off")

    plt.subplot(3, 4, 7)
    plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
    plt.title("Blurred")
    plt.axis("off")

    plt.subplot(3, 4, 8)
    plt.imshow(cv2.cvtColor(bright, cv2.COLOR_BGR2RGB))
    plt.title("Brightness + Contrast")
    plt.axis("off")

    plt.subplot(3, 4, 9)
    plt.imshow(edges, cmap="gray")
    plt.title("Edge Detection")
    plt.axis("off")

    plt.subplot(3, 4, 10)
    plt.imshow(cv2.cvtColor(text_image, cv2.COLOR_BGR2RGB))
    plt.title("Text Added")
    plt.axis("off")

    plt.subplot(3, 4, 11)
    plt.imshow(cv2.cvtColor(bordered, cv2.COLOR_BGR2RGB))
    plt.title("Border Added")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

   
    cv2.imwrite("01_resized.jpg", resized)
    cv2.imwrite("02_cropped.jpg", cropped)
    cv2.imwrite("03_rotated.jpg", rotated)
    cv2.imwrite("04_flipped.jpg", flipped)
    cv2.imwrite("05_grayscale.jpg", grayscale)
    cv2.imwrite("06_blurred.jpg", blurred)
    cv2.imwrite("07_brightness.jpg", bright)
    cv2.imwrite("08_edges.jpg", edges)
    cv2.imwrite("09_text_added.jpg", text_image)
    cv2.imwrite("10_bordered.jpg", bordered)

   
    print("IMAGE EDITING COMPLETED!")
   
    print("10 edited images have been saved.")
```
